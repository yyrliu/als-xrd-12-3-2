```python

def read_log_file_and_export(log_txt: Path, out_path: Path, headers=None):

# find line number of the string "DATA"
    data_line_num = None
    with open(log_txt, 'r') as f:
        for i, line in enumerate(f):
            if 'DATA' in line:
                data_line_num = i
                break

    if data_line_num is None:
        raise ValueError(f'Could not find "DATA" in the log file `{str(log_txt)}`.')

    # read the data into a pandas dataframe
    df = pd.read_csv(log_txt, skiprows=data_line_num + 1, sep='\t')

    df = df[df['Pilatus'].notna()]

    # frame_number is xxx.tiff, extract xxx as an integer
    df['frame_number'] = df['Pilatus'].str.extract(r'(\d+)\.tiff').astype(int)

    df.to_csv(out_path, index=False, mode='a')

    print(f'Saved extracted timestamps to {out_path.name}')
```
